```python
import numpy as np
import pandas as pd
```

```python
df  =  pd.read_csv('customer.csv')
```

```python
df.sample(5)

#Categorical_cols --> gender	review	education	purchased 

# gender	--> Nominal Categorical  --> One Hot Encoder
# review    --> Ordinal Categorical  --> Ordinal Encoder
# education --> Ordinal Categorical  --> Ordinal Encoder
# purchased --> Nominal Categorical --> Label Encoder

```

```python
df = df.iloc[: ,2:]  # All Rows & 2,3,4 Cols
```

```python
df.head(15)
```

# Train Test Split  

```python
from sklearn.model_selection import train_test_split
X_train, X_test , y_train, y_test = train_test_split(df.iloc[:,0:2], df.iloc[:,-1],test_size = 0.2)
```

```python
X_train
```

# Ordinal Encoding

## Converts Categorical data into ordered Numeric Data

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories = [['Poor', 'Average','Good'],['School','UG','PG']])

## If we don't give the proper ordered categories it will select randomly 
```

```python
oe.fit(X_train)
```

#  Data Transformation

```python
X_train = oe.transform(X_train)
X_test = oe.transform(X_test)
```

```python
X_train
```

```python
oe.categories_
```

# Label Encoding

## Convert O/P labeled data into Numerical Data

- Encode target labels with value between 0 and n_classes-1.

- This transformer should be used to encode target values, i.e. y, and not the input X

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le  = LabelEncoder()

## here we can't give order to categories(like Ordinal) , it will give by itself
#
```

```python
le.fit(y_train)
```

```python
le.classes_
```

```python
y_train = le.transform(y_train)
y_test = le.transform(y_test)
```
