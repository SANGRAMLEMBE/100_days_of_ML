```python
## to fetch data from differnt pages

final = pd.DataFrame()

for j in range(1,31):

    URL= 'https://www.ambitionbox.com/list-of-companies?page={}'.format(j)
    ## To Fetch HTML/ css code from website
    headers = { 'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'} 
    webpage = requests.get(URL,headers=headers).text

    soup =  BeautifulSoup(webpage,'lxml') 

    company = soup.find_all('div', class_= 'companyCardWrapper')

    
    names = []
    rating = []
    reviews = []
    company_data = []
    
    
    for i in company:
        names.append(i.find('h2').text.strip())
        reviews.append(i.find_all( class_ = 'companyCardWrapper__companyRatingCount')[0].text.strip())
        rating.append(i.find('div',class_ = 'rating_text').text.strip())
        company_data.append(i.find('span',class_ = 'companyCardWrapper__interLinking').text.strip())
    
        # OR
    
        ##rating.append(i.find( class_ = 'companyCardWrapper__companyRatingCount').text.strip())
    
    
        # find  --> for only 1 (we have only 1 h2 )
        # find_all --> when many attributes are present in class (rating, location,..)
        # [0] --> only 1 value is present 
        # .text.strip() --> to extract the data(rating/ name)
    
    
    d = {'names': names, 'rating' :rating, 'reviews':reviews, 'company_data':company_data}
    df = pd.DataFrame(d)

    final = pd.concat([final, df], ignore_index=True)
    

    
```
